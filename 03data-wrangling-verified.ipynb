### TASK 2 — Number and occurrence of each orbit


```python
# Task 2
# Calculate the number and occurrence of each orbit, excluding GTO as instructed
df[df['Orbit'] != 'GTO']['Orbit'].value_counts()
```

### TASK 3 — Number and occurrence of landing outcomes


```python
# Task 3
# Calculate the number and occurrence of each landing outcome
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1,3,5,6,7]])
bad_outcomes
```

### TASK 4 — Create the landing outcome training label


```python
# Task 4
# Create the training label: 0 for unsuccessful landings, 1 otherwise
landing_class = []
for outcome in df['Outcome']:
    if outcome in bad_outcomes:
        landing_class.append(0)
    else:
        landing_class.append(1)

df['Class'] = landing_class
df[['Outcome', 'Class']].head(8)
```

```python
df["Class"].mean()
```
