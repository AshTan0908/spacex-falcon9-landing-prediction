```python
import csv, sqlite3
import prettytable
prettytable.DEFAULT = 'DEFAULT'

con = sqlite3.connect("my_data1.db")
cur = con.cursor()
```

```python
import pandas as pd
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False,method="multi")
```

```python
#DROP THE TABLE IF EXISTS

%sql DROP TABLE IF EXISTS SPACEXTABLE;

cur.execute("create table SPACEXTABLE as select * from SPACEXTBL where Date is not null")
```

```python
%sql SELECT name FROM sqlite_master WHERE type='table';
```

```python
%sql SELECT * FROM SPACEXTABLE LIMIT 10
```

```python
#Task 1
%sql SELECT DISTINCT Launch_Site FROM SPACEXTABLE;
```

```python
#Task 2
%sql SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5;
```

```python
#Task 3
%sql SELECT SUM(PAYLOAD_MASS__KG_) FROM SPACEXTABLE WHERE Customer LIKE '%NASA%';
```

```python
#Task 4
%sql SELECT AVG(PAYLOAD_MASS__KG_) FROM SPACEXTABLE WHERE Booster_Version LIKE 'F9 v1.1'
```

```python
#Task 5
%sql SELECT MIN(Date) FROM SPACEXTABLE WHERE Landing_Outcome LIKE 'Success';
```

```python
#Task 6
%sql SELECT * FROM SPACEXTABLE WHERE Mission_Outcome LIKE 'Success' and PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000;
```

```python
#Task 7
%sql SELECT Mission_Outcome, COUNT(*) AS Total_Count FROM SPACEXTABLE GROUP BY Mission_Outcome;
```

```python
#Task 8 
%sql SELECT Booster_Version FROM SPACEXTABLE WHERE Payload_Mass__kg_ = (SELECT MAX(Payload_Mass__kg_) FROM SPACEXTABLE);
```

```python
#Task 9
%sql SELECT strftime('%m', Date) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE Date LIKE '2015-%' AND Landing_Outcome = 'Failure (drone ship)';
```

```python
#Task 10
%sql SELECT Landing_Outcome, COUNT(*) AS Count FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Count DESC; 
```
